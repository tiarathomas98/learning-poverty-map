```python
import pandas as pd

# 1. State-level public health dataset
data = {
    'State': ['Tamil Nadu', 'Gujarat', 'Maharashtra', 'Kerala', 'Madhya Pradesh', 'Uttar Pradesh'],
    'Per_Capita_Spend': [3250, 3180, 3400, 3600, 2100, 1950],
    'SDG_Health_Score': [78.5, 64.2, 74.0, 82.1, 52.3, 48.6],
    'Facility_Op_Rate': [92.4, 71.0, 85.0, 96.2, 58.0, 51.2],
    'Stockout_Variance': [4.2, 18.5, 8.1, 2.5, 28.4, 32.1]
}

df = pd.DataFrame(data)

# 2. Min-Max Normalization function (inverting negative metrics like stockouts)
def normalize(series, invert=False):
    min_val = series.min()
    max_val = series.max()
    norm = (series - min_val) / (max_val - min_val)
    return (1 - norm) if invert else norm

df['Norm_SDG'] = normalize(df['SDG_Health_Score'])
df['Norm_Op_Rate'] = normalize(df['Facility_Op_Rate'])
df['Norm_Stockout'] = normalize(df['Stockout_Variance'], invert=True)

# 3. Compute Composite Public Health Delivery Index (Scale 0-100)
df['Composite_Delivery_Index'] = ((df['Norm_SDG'] + df['Norm_Op_Rate'] + df['Norm_Stockout']) / 3) * 100

# 4. Display ranked results
ranked_df = df[['State', 'Per_Capita_Spend', 'Composite_Delivery_Index']].sort_values(by='Composite_Delivery_Index', ascending=False)
print(ranked_df.to_string(index=False))
```
